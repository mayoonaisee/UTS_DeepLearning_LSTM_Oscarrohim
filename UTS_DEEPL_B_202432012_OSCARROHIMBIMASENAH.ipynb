  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432012_Oscarrohim Bimasenah
# Mengimpor pustaka untuk manipulasi array dan data tabular
import numpy as np
import pandas as pd
# Mengimpor pustaka untuk visualisasi grafik
import matplotlib.pyplot as plt
# Mengimpor pustaka utama PyTorch untuk Deep Learning
import torch
import torch.nn as nn
import torch.optim as optim
# Mengimpor modul untuk normalisasi data deret waktu
from sklearn.preprocessing import MinMaxScaler
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432012_Oscarrohim Bimasenah
# 1. Memuat dataset saham Apple dari file CSV menggunakan pandas
df = pd.read_csv('/content/sample_data/AAPL.csv')
# Mengambil kolom 'Close' (harga penutupan saham) dan mengubahnya menjadi array 2D
data = df['Close'].values.reshape(-1, 1)

# 2. Inisiasi scaler untuk menormalkan data ke rentang 0 hingga 1 agar model stabil
scaler = MinMaxScaler(feature_range=(0, 1))
# Menerapkan normalisasi pada data asli
data_scaled = scaler.fit_transform(data)

# Fungsi untuk membentuk sliding window (urutan sekuensial)
def create_sequences(data, seq_length):
    xs, ys = [], [] # List kosong untuk fitur (X) dan target (y)
    # Looping sepanjang data dikurangi panjang sekuens
    for i in range(len(data) - seq_length):
        # Mengambil data dari indeks i hingga i+seq_length sebagai input
        x = data[i:(i + seq_length)]
        # Mengambil 1 langkah waktu setelahnya sebagai target prediksi
        y = data[i + seq_length]
        xs.append(x)
        ys.append(y)
    return np.array(xs), np.array(ys)

seq_length = 60 # Menggunakan data 60 hari perdagangan terakhir untuk memprediksi hari berikutnya
X, y = create_sequences(data_scaled, seq_length)

# Membagi 80% data untuk training dan 20% untuk testing
train_size = int(len(X) * 0.8)
X_train, y_train = X[:train_size], y[:train_size]
X_test, y_test = X[train_size:], y[train_size:]

# 3. Konversi array numpy menjadi PyTorch Tensor dengan tipe data float32
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432012_Oscarrohim Bimasenah
# Mendefinisikan kelas model yang mewarisi modul nn.Module PyTorch
class SequenceModel(nn.Module):
    # Konstruktor model dengan parameter input, hidden layer, dan output
    def __init__(self, input_size=1, hidden_size=50, num_layers=1, output_size=1):
        super(SequenceModel, self).__init__()
        # Mendefinisikan layer LSTM dengan parameter batch_first=True agar dimensi pertama adalah batch
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        # Mendefinisikan layer Linear (Fully Connected) untuk memetakan output LSTM ke 1 nilai prediksi
        self.linear = nn.Linear(hidden_size, output_size)

    # Fungsi forward propagation untuk alur maju data
    def forward(self, x):
        # Memasukkan input ke layer LSTM; menghasilkan output dan hidden state
        out, _ = self.lstm(x)
        # Mengambil output hanya dari langkah waktu (time step) terakhir
        out = self.linear(out[:, -1, :])
        return out # Mengembalikan hasil prediksi
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432012_Oscarrohim Bimasenah
# 1. Inisiasi hyperparameter
epochs = 50 # Jumlah iterasi pelatihan keseluruhan dataset
learning_rate = 0.01 # Kecepatan pembaruan bobot model

# 2. Inisiasi objek model jaringan LSTM
model = SequenceModel()
# Menggunakan Mean Squared Error (MSE) karena ini adalah tugas regresi/forecasting
criterion = nn.MSELoss()
# Menggunakan optimizer Adam untuk mengoptimalkan parameter model
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. List kosong untuk menyimpan nilai loss di setiap epoch
train_losses = []

# 4. Training Loop (Siklus Pelatihan)
for epoch in range(epochs):
    model.train() # Mengatur model ke mode pelatihan
    optimizer.zero_grad() # Mengosongkan sisa gradien dari iterasi sebelumnya

    # Forward pass: Menghitung prediksi model berdasarkan X_train
    output = model(X_train)
    # Menghitung nilai error (loss) antara prediksi dan y_train aktual
    loss = criterion(output, y_train)

    # Backward pass: Menghitung gradien dari loss (Backpropagation)
    loss.backward()
    # Memperbarui bobot model berdasarkan gradien
    optimizer.step()

    # Menyimpan nilai loss ke dalam list
    train_losses.append(loss.item())

    # Mencetak log progres setiap 10 epoch
    if (epoch+1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.6f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432012_Oscarrohim Bimasenah
# 1. Plot Grafik Penurunan Loss Training
plt.figure(figsize=(10, 4))
# Memplot array train_losses untuk melihat tren error
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Evaluasi model pada data uji (X_test)
model.eval() # Mengubah model ke mode evaluasi
with torch.no_grad(): # Menonaktifkan perhitungan gradien untuk menghemat memori
    test_predictions = model(X_test)

# Denormalisasi nilai prediksi dan aktual kembali ke skala harga saham asli
test_predictions = scaler.inverse_transform(test_predictions.numpy())
y_test_actual = scaler.inverse_transform(y_test.numpy())

# Plot Grafik Perbandingan Data Aktual vs Hasil Prediksi
plt.figure(figsize=(12, 5))
# Menampilkan 200 data uji pertama agar grafik perbandingan terlihat jelas
plt.plot(y_test_actual[:200], color='#114F6A', label='Harga Aktual (USD)', linewidth=1.5)
plt.plot(test_predictions[:200], color='#1FA4B1', label='Prediksi LSTM', linewidth=1.5, linestyle='--')
plt.title('Perbandingan Harga Saham Apple Aktual vs Prediksi (200 Data Uji Pertama)')
plt.xlabel('Waktu (Hari Perdagangan)')
plt.ylabel('Harga Saham (USD)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
